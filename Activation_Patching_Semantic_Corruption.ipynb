```python
import torch
import numpy as np

# 1. Convert logits to probability distributions
p = torch.softmax(logits[0, -1, :], dim=-1)
q = torch.softmax(corrupted_logits[0, -1, :], dim=-1)

# 2. Total Variation (TV) Distance: 0.5 * sum(|p_i - q_i|)
tv_distance = 0.5 * torch.sum(torch.abs(p - q)).item()

# 3. KL Divergence: sum(p_i * log(p_i / q_i))
kl_p_q = torch.sum(p * torch.log(p / (q + 1e-12))).item()
kl_q_p = torch.sum(q * torch.log(q / (p + 1e-12))).item()

# 4. Jensen-Shannon (JS) Divergence: 0.5 * KL(P || M) + 0.5 * KL(Q || M) where M = 0.5 * (P + Q)
m = 0.5 * (p + q)
kl_p_m = torch.sum(p * torch.log(p / (m + 1e-12)))
kl_q_m = torch.sum(q * torch.log(q / (m + 1e-12)))
js_divergence = (0.5 * kl_p_m + 0.5 * kl_q_m).item()
js_distance = np.sqrt(js_divergence) if js_divergence >= 0 else 0.0

# Print the results
print(f"--- Next-Token Distribution Change Metrics ---")
print(f"Total Variation Distance: {tv_distance:.4f}")
print(f"KL Divergence D_KL(Clean || Corrupted): {kl_p_q:.4f} nats")
print(f"KL Divergence D_KL(Corrupted || Clean): {kl_q_p:.4f} nats")
print(f"Jensen-Shannon Divergence: {js_divergence:.4f}")
print(f"Jensen-Shannon Distance: {js_distance:.4f}")
```

```python
import torch
import transformer_lens
import transformer_lens.utils as utils
from transformer_lens import HookedTransformer
import plotly.express as px


# 1. Load the GPT2-Medium model
device = "cuda" if torch.cuda.is_available() else "cpu"
model = HookedTransformer.from_pretrained("gpt2-medium", device=device)

# 2. Define the prompts and tokenize
clean_prompt = "My dog is a good"
corrupt_prompt = "My cat is a good"

# We evaluate the logit outputs at the final token position (' good')
final_token_pos = -1

# Get the precise token IDs for our target outputs
dog_token_id = model.to_single_token(" dog")
cat_token_id = model.to_single_token(" cat")

# 3. Run the model and capture all internal activations (cache)
# We calculate DLA on the clean prompt run to see what normally happens,
# or you can run it on your patched activation state.
logits, cache = model.run_with_cache(clean_prompt)

# 4. Extract the Unembedding weights (W_U) for our target tokens
# We want the direction in the residual stream that corresponds to: (cat_logit - dog_logit)
W_U = model.W_U
dla_direction = W_U[:, cat_token_id] - W_U[:, dog_token_id]

# 5. Calculate Direct Logit Attribution for all Attention Heads
# GPT2-Medium has 24 layers and 16 heads per layer
per_head_dla = torch.zeros((model.cfg.n_layers, model.cfg.n_heads))

for layer in range(model.cfg.n_layers):
    # Get the output of every head in this layer at the final token position
    # Shape: [batch, position, head_index, d_model]
    head_outputs = cache[utils.get_act_name("z", layer)][0, final_token_pos, :, :]

    # Project the head outputs through the output projection matrix W_O
    # Shape: [head_index, d_model]
    head_projections = torch.einsum("hd,hdm->hm", head_outputs, model.blocks[layer].attn.W_O)

    # Compute the dot product with our logit difference direction
    # This tells us how much each head contributes to (cat_logit - dog_logit)
    for head in range(model.cfg.n_heads):
        per_head_dla[layer, head] = torch.dot(head_projections[head], dla_direction).item()

# 6. Plot the DLA Heatmap using Plotly
fig = px.imshow(
    per_head_dla.cpu().numpy(),
    labels=dict(x="Attention Head", y="Layer", color="Logit Attribution Diff"),
    x=[f"H{h}" for h in range(model.cfg.n_heads)],
    y=[f"L{l}" for l in range(model.cfg.n_layers)],
    title="Direct Logit Attribution Difference (cat - dog) per Head at final token position",
    color_continuous_scale="RdBu",  # Red = favors 'cat', Blue = favors 'dog'
    color_continuous_midpoint=0.0
)

fig.update_layout(width=700, height=600)
fig.show()
```

#### Activation Patching

**Direct Logit Attribution (DLA)**

```python

import torch
import transformer_lens
import transformer_lens.utils as utils
from transformer_lens import HookedTransformer
import plotly.express as px

# 1. Load GPT-2 Medium
device = "cuda" if torch.cuda.is_available() else "cpu"
model = HookedTransformer.from_pretrained(
    "gpt2-medium",
    device=device
)
model.eval()

# 2. Define clean and corrupt prompts
clean_prompt = "My dog is a good"
corrupt_prompt = "My cat is a good"
final_token_pos = -1

dog_token_id = model.to_single_token(" dog")
cat_token_id = model.to_single_token(" cat")

# 3. Cache activations from both runs
with torch.no_grad():
    clean_logits, clean_cache = model.run_with_cache(clean_prompt)
    corrupt_logits = model(corrupt_prompt)

# 4. Define actual logit difference: dog - cat
def logit_diff(logits):
    return (
        logits[0, final_token_pos, dog_token_id]
        - logits[0, final_token_pos, cat_token_id]
    )

corrupt_diff = logit_diff(corrupt_logits).item()
clean_diff = logit_diff(clean_logits).item()

print(f"Corrupt dog-cat logit difference: {corrupt_diff:.4f}")
print(f"Clean dog-cat logit difference:   {clean_diff:.4f}")

# 5. Patch each head's z activation at the final position
# z shape: [batch, position, head, d_head]
results = torch.zeros(
    model.cfg.n_layers,
    model.cfg.n_heads
)

for layer in range(model.cfg.n_layers):
    hook_name = utils.get_act_name("z", layer)

    for head in range(model.cfg.n_heads):

        clean_head_z = clean_cache[hook_name][
            0, final_token_pos, head, :
        ].detach().clone()

        def patch_head(z, hook, head_idx=head, clean_z=clean_head_z):
            z[:, final_token_pos, head_idx, :] = clean_z
            return z

        with torch.no_grad():
            patched_logits = model.run_with_hooks(
                corrupt_prompt,
                fwd_hooks=[(hook_name, patch_head)]
            )

        patched_diff = logit_diff(patched_logits).item()
        results[layer, head] = patched_diff - corrupt_diff

# 6. Plot change in the actual logit difference
fig = px.imshow(
    results.cpu().numpy(),
    labels={
        "x": "Attention Head",
        "y": "Layer",
        "color": "Delta Logit Difference"
    },
    x=[f"H{h}" for h in range(model.cfg.n_heads)],
    y=[f"L{l}" for l in range(model.cfg.n_layers)],
    title="Activation Patching: Change in (dog - cat) Logit Difference",
    color_continuous_scale="RdBu",
    color_continuous_midpoint=0.0
)

fig.update_layout(width=800, height=650)
fig.show()
```

### Complete Activation Patching Analysis

Now, let's expand our activation patching beyond individual attention heads to analyze the entire network architecture systematically. We will patch the following components at the final token position (`-1`) for every layer:
* **Residual Stream (`resid_pre`)**
* **Attention Layer Output (`attn_out`)**
* **MLP Layer Output (`mlp_out`)**

This will help isolate whether the causal contribution originates from Attention, MLPs, or their accumulation in the residual stream.

```python
import torch
import pandas as pd
import plotly.graph_objects as go
import transformer_lens.utilities as utils

# 1. Setup containers to hold patching results
layers = list(range(model.cfg.n_layers))
results_resid = []
results_attn = []
results_mlp = []

# 2. Define clean & corrupt difference targets
dog_token_id = model.to_single_token(" dog")
cat_token_id = model.to_single_token(" cat")

def get_logit_diff(logits):
    return logits[0, -1, dog_token_id] - logits[0, -1, cat_token_id]

# Calculate baseline logit differences
with torch.no_grad():
    clean_logits, clean_cache = model.run_with_cache(clean_prompt)
    corrupt_logits, _ = model.run_with_cache(corrupted_prompt)

clean_diff = get_logit_diff(clean_logits).item()
corrupt_diff = get_logit_diff(corrupt_logits).item()

# 3. Perform the sweep over all layers
for layer in layers:
    # Define Hook Points
    resid_hook = utils.get_act_name("resid_pre", layer)
    attn_hook = utils.get_act_name("attn_out", layer)
    mlp_hook = utils.get_act_name("mlp_out", layer)

    # Function generator for patching
    def make_patch_hook(clean_activation):
        def patch_hook(corrupt_activation, hook):
            corrupt_activation[:, -1, :] = clean_activation[:, -1, :]
            return corrupt_activation
        return patch_hook

    # Patch Residual Stream
    with torch.no_grad():
        patched_resid_logits = model.run_with_hooks(
            corrupted_prompt,
            fwd_hooks=[(resid_hook, make_patch_hook(clean_cache[resid_hook]))]
        )
        results_resid.append(get_logit_diff(patched_resid_logits).item() - corrupt_diff)

    # Patch Attention Output
    with torch.no_grad():
        patched_attn_logits = model.run_with_hooks(
            corrupted_prompt,
            fwd_hooks=[(attn_hook, make_patch_hook(clean_cache[attn_hook]))]
        )
        results_attn.append(get_logit_diff(patched_attn_logits).item() - corrupt_diff)

    # Patch MLP Output
    with torch.no_grad():
        patched_mlp_logits = model.run_with_hooks(
            corrupted_prompt,
            fwd_hooks=[(mlp_hook, make_patch_hook(clean_cache[mlp_hook]))]
        )
        results_mlp.append(get_logit_diff(patched_mlp_logits).item() - corrupt_diff)

# 4. Visualize results
fig = go.Figure()
fig.add_trace(go.Scatter(x=layers, y=results_resid, name="Residual Stream (resid_pre)", mode="lines+markers", line=dict(color="#1f77b4", width=3)))
fig.add_trace(go.Scatter(x=layers, y=results_attn, name="Attention Output (attn_out)", mode="lines+markers", line=dict(color="#ff7f0e", width=2)))
fig.add_trace(go.Scatter(x=layers, y=results_mlp, name="MLP Output (mlp_out)", mode="lines+markers", line=dict(color="#2ca02c", width=2)))

fig.update_layout(
    title="Complete Activation Patching by Layer Component",
    xaxis_title="Layer",
    yaxis_title="Delta Logit Difference (dog - cat)",
    legend_title="Component patched",
    width=1000,
    height=550,
    grid=dict(rows=1, columns=1)
)
fig.show()
```

### Findings: When the Final Token Becomes Causally Important in GPT-2 Medium

This document summarizes our exploratory study on how the representation of the final token evolves and when it begins to exert a causal influence on next-token prediction in GPT-2 Medium (24 layers, 16 heads).

#### 1. Experimental Setup
* **Clean Prompt:** `"My dog is a good"` (Expected target: `" dog"` / high probability of canine continuation)
* **Corrupt Prompt:** `"My cat is a good"` (Expected target: `" cat"` / high probability of feline continuation)
* **Methodology:** We performed **Activation Patching** on the key attention outputs ($z$ activations) of every individual head at the final token position (`" good"`), swapping activations from the clean run into the corrupt run.
* **Metric:** We calculated the $\Delta$ Logit Difference ($[\text{Logit}(\text{dog}) - \text{Logit}(\text{cat})]_{\text{patched}} - [\text{Logit}(\text{dog}) - \text{Logit}(\text{cat})]_{\text{corrupt}}$).

#### 2. Key Phenomenon: The Causal Transition Window
By filtering out heads with negligible logit impact ($|\Delta| < 0.05$) or where the top predictions did not diverge, we observed a distinct developmental trajectory across the network's layers:

* **Early Layers (L0 to L10) - The Silent Phase:**
  * Patching attention heads in these layers yielded virtually **zero change** in the final logit difference.
  * This indicates that the early layers are performing local contextualization and representation-building but are not yet computing the direct causal features used to resolve the final target prediction.
  
* **The Causal Onset - Layer 11 (L11):**
  * **L11** acts as the sharp boundary where the token's predictive identity first becomes causally active in the attention stream.
  * This is the earliest point where patching a single head's output begins to measurably swing the final logit difference and alter predicted tokens (e.g., resolving conflicting states like `resil/henko`).

* **Late Layers (L12 to L23) - The Direct Attributors:**
  * From Layer 11 onwards, we observe specific, highly specialized heads (e.g., in Layers 13, 14, 15, and late 20s) executing the final token prediction routing.
  * These heads actively translate the contextual features built up in the first half of the model directly into logit adjustments favoring the clean target `" dog"` over `" cat"`.
